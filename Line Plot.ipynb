`````python
import pandas as pd

# Create sample data
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Product_A": [12000, 15000, 14000, 18000, 20000, 22000],
    "Product_B": [10000, 13000, 12500, 16000, 17000, 19000],
    "Product_C": [8000, 9000, 9500, 11000, 13000, 15000],
    "Product_D": [6000, 7000, 8500, 9000, 10500, 12000]
}

df = pd.DataFrame(data)
print(df)
`````

`````python
plt.plot(df["Month"], df["Product_A"], label = "Product_A", marker = 's', color = 'pink', linewidth = 2)
plt.plot(df["Month"], df["Product_B"], label = "Product_B", marker = 's', color = 'darkblue', linewidth = 2)
plt.plot(df["Month"], df["Product_C"], label = "Product_C", marker = 's', color = 'yellow', linewidth = 2)
plt.plot(df["Month"], df["Product_D"], label = "Product_D", marker = 's', color = 'green', linewidth = 2)

plt.legend()
`````

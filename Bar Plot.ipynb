## **Bar Plot**
------

```python
import pandas as pd

data = {
    "Product": ["Product A", "Product B", "Product C", "Product D"],
    "2024_Sales": [22000, 19000, 15000, 12000],
    "2025_Sales": [25000, 21000, 17000, 14000]
}

df = pd.DataFrame(data)
df
```

```python
import matplotlib.pyplot as plt

colors = ["pink", "blue", "green", "black"]
plt.bar(df["Product"], df["2024_Sales"])
```

```python
import matplotlib.pyplot as plt

colors = ["pink", "darkblue", "green", "black"]
plt.bar(df["Product"], df["2024_Sales"], color = colors)
```

## **Count Plot**

```python
import pandas as pd

# Create sample categorical data
df = pd.DataFrame({
    "Gender": ["Male", "Female", "Female", "Male", "Male", "Female"],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR"]
})
df
```

```python
count = df["Department"].value_counts()
```

```python
import matplotlib.pyplot as plt

plt.bar(count.index, count.values, color = ['pink', 'blue', 'green'])
```

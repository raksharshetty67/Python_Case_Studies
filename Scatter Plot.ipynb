## **Scatter Plot**
------------

```python
import pandas as pd

# Create sample data
df = pd.DataFrame({
    "Age": [22, 25, 30, 28, 35, 40],
    "Salary": [25000, 30000, 45000, 38000, 60000, 75000],
    "Experience": [1, 2, 5, 4, 8, 12],
    "Working_Hours": [8, 9, 10, 8, 9, 11]
})
df
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["Age"], df["Salary"])
```

```python
plt.scatter(df["Age"], df["Salary"], color = 'salmon')
```

```python
plt.scatter(df["Age"], df["Salary"])
plt.title("Scatter Plot")
plt.xlabel("Age")
plt.ylabel("Salary")
```
